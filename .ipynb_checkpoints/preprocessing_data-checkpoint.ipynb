```python
import pandas as pd
df = pd.read_csv('data/eurojackpot_2012_2025_master.csv')
df.head()
```

```python
print(df.dtypes)
```

```python
df['short_date'] = pd.to_datetime(df['short_date'], format='%d.%m.%Y')
df.tail()
```

```python
df.info()
```

```python
# With null support:
df['jackpot'] = (
    df['jackpot']
    .str.replace('€', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)            # allows null values
)
```

```python
# With null support:
df['prize fund'] = (
    df['prize fund']
    .str.replace('€', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype('Int64')            # allows null values
)
```

```python
df['prize / winner'] = (
    df['prize / winner']
    .str.replace('€', '', regex=False)
    .str.replace(',', '', regex=False)
)
df['prize / winner'] = pd.to_numeric(df['prize / winner'], errors='coerce')
```

```python
df
```

```python
df['main_numbers'] = df['main_numbers'].apply(lambda x: [int(n) for n in x.split()])
df['euro_numbers'] = df['euro_numbers'].apply(lambda x: [int(n) for n in x.split()])
```

```python
row = df.iloc[7222]
print(row)
```

```python
df.info()
```

```python
df['# of winners'] = (
    df['# of winners']
    .str.replace(',', '', regex=False)
)

df['# of winners'] = pd.to_numeric(df['# of winners'], errors='coerce').astype('Int64')
```

```python
df
```

```python
df["Match"] = df["Match"].astype(str)
```
