## Jupyter Notebook starter code 
- snippet to combine all yearly CSV files from the data folder into one master dataset

```python
import pandas as pd
import os

# Folder containing your yearly CSVs
data_folder = 'data'

# Years to combine
years = range(2012, 2026)

# Load and concatenate CSVs into one DataFrame
df_list = []
for year in years:
    file_path = os.path.join(data_folder, f'eurojackpot_{year}_all_draws.csv')
    if os.path.exists(file_path):
        df_year = pd.read_csv(file_path)
        df_list.append(df_year)
    else:
        print(f"File missing: {file_path}")

master_df = pd.concat(df_list, ignore_index=True)
print("Combined dataset info:")
print(master_df.info())

# Preview first rows
print("\nFirst 5 rows:")
print(master_df.head())

# Basic stats summary, e.g., jackpot distribution
print("\nJackpot value counts:")
print(master_df['jackpot'].value_counts().head())

# Save combined master dataset
master_df.to_csv(os.path.join(data_folder, 'eurojackpot_2012_2025_master.csv'), index=False)
print("\nMaster dataset saved to 'data/eurojackpot_2012_2025_master.csv'")
```
